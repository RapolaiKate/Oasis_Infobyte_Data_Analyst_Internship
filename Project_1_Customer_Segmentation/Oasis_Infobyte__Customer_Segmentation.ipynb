```python
monetary = df_clean.groupby("CustomerID")["TotalAmount"].sum()

monetary.head()
```

```python
frequency = df_clean.groupby("CustomerID")["InvoiceNo"].nunique()
```

```python
frequency.head()
```

```python
monetary = df_clean.groupby("CustomerID")["TotalAmount"].sum()
```

```python
rfm = pd.concat(
    [recency, frequency, monetary],
    axis=1
)

rfm.columns = ["Recency", "Frequency", "Monetary"]

rfm.head()
```

```python
rfm["R_Score"] = pd.qcut(
    rfm["Recency"],
    5,
    labels=[5, 4, 3, 2, 1]
)
```

```python
rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
)
```

```python
rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
)
```

```python
rfm.head()
```

```python
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(int)
    + rfm["F_Score"].astype(int)
    + rfm["M_Score"].astype(int)
)

rfm.head()
```

```python
def segment_customer(row):
    score = row["RFM_Score"]
    
    if score >= 13:
        return "Champions"
    elif score >= 10:
        return "Loyal Customers"
    elif score >= 7:
        return "Potential Loyalists"
    elif score >= 5:
        return "At Risk"
    else:
        return "Lost Customers"

rfm["Segment"] = rfm.apply(segment_customer, axis=1)

rfm.head()
```

```python
segment_counts = rfm["Segment"].value_counts()

segment_counts
```

```python
segment_percentage = (
    rfm["Segment"].value_counts(normalize=True) * 100
).round(2)

segment_percentage
```

```python
segment_summary = pd.DataFrame({
    "Customer_Count": segment_counts,
    "Percentage": segment_percentage
})

segment_summary
```

# Business Insights and Recommendations

The RFM analysis segmented 4,372 customers into five groups based on their purchasing behavior: Potential Loyalists, Loyal Customers, Champions, At Risk, and Lost Customers.

## Key Findings

### 1. Potential Loyalists

* **Customers:** 1,114
* **Percentage:** 25.48%

Potential Loyalists represent the largest customer segment. These customers show potential for stronger engagement and loyalty.

**Recommendation:** The business should use personalized promotions, loyalty rewards, product recommendations, and targeted follow-up campaigns to encourage repeat purchases and move these customers toward the Loyal Customers and Champions segments.

### 2. Loyal Customers

* **Customers:** 990
* **Percentage:** 22.64%

Loyal Customers demonstrate strong purchasing relationships with the business and represent an important customer base.

**Recommendation:** The business should maintain these relationships through loyalty programs, exclusive offers, personalized communication, and early access to new products.

### 3. Champions

* **Customers:** 949
* **Percentage:** 21.71%

Champions are the strongest customer segment according to the RFM analysis. They demonstrate valuable purchasing behavior through recent purchases, frequent transactions, and higher spending.

**Recommendation:** The business should prioritize retaining Champions by providing VIP rewards, exclusive benefits, personalized experiences, and early access to products or promotions.

### 4. At Risk

* **Customers:** 782
* **Percentage:** 17.89%

At Risk customers represent a significant group whose purchasing activity indicates that they may become inactive.

**Recommendation:** The business should use targeted re-engagement campaigns, personalized discounts, reminders, and special incentives to encourage these customers to return.

### 5. Lost Customers

* **Customers:** 537
* **Percentage:** 12.28%

Lost Customers have low engagement and may require specific win-back strategies.

**Recommendation:** The business should test targeted win-back campaigns and evaluate previous customer value before allocating significant resources to reactivation.

## Overall Business Insights

The analysis identified 4,372 customers across five customer segments. Potential Loyalists form the largest segment at 25.48%, representing a major opportunity to increase customer loyalty and repeat purchases.

Loyal Customers and Champions together account for 44.35% of the customer base, providing the business with a strong group of customers to retain and develop.

At Risk and Lost Customers together account for 30.17% of the customer base. This indicates that customer retention and re-engagement should be important business priorities.

## Strategic Priorities

1. Convert Potential Loyalists into Loyal Customers and Champions.
2. Protect and reward Champions and Loyal Customers.
3. Re-engage At Risk customers before they become inactive.
4. Test targeted win-back campaigns for Lost Customers.
5. Use RFM segmentation to support personalized marketing and customer retention strategies.


# Project Conclusion

The Customer Segmentation analysis successfully used RFM (Recency, Frequency, and Monetary) analysis to understand customer purchasing behavior and divide 4,372 customers into meaningful segments.

The analysis identified five customer groups: Potential Loyalists, Loyal Customers, Champions, At Risk, and Lost Customers. Potential Loyalists represented the largest segment at 25.48%, while Loyal Customers and Champions together accounted for 44.35% of the customer base.

The results also showed that 30.17% of customers were classified as At Risk or Lost Customers, highlighting the importance of customer retention and re-engagement strategies.

Overall, the RFM analysis provides the business with a data-driven approach to customer relationship management. By targeting each customer segment with appropriate strategies, the business can strengthen customer loyalty, improve retention, encourage repeat purchases, and identify opportunities for future growth.
