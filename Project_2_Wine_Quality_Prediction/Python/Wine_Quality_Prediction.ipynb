## DATA CLEANING

```python
# Select numerical columns
numeric_df = df.select_dtypes(include='number')

# Calculate Q1, Q3 and IQR
Q1 = numeric_df.quantile(0.25)
Q3 = numeric_df.quantile(0.75)
IQR = Q3 - Q1

# Count potential outliers
outliers = ((numeric_df < (Q1 - 1.5 * IQR)) |
            (numeric_df > (Q3 + 1.5 * IQR))).sum()

outliers
```

## EXPLORATOTY DATA ANALYSIS (EDA)

```python
# Calculate the number of wines for each quality score

quality_counts = df['quality'].value_counts().sort_index()
quality_counts
```

```python
# Import matplotlib for data visualization

import matplotlib.pyplot as plt
```

```python
# Visualize the distribution of wine quality scores

df['quality'].value_counts().sort_index().plot(kind='bar', figsize=(8,5))

plt.title('Distribution of Wine Quality')
plt.xlabel('Quality Score')
plt.ylabel('Number of Wines')
plt.xticks(rotation=0)
plt.show()
```

```python
# Visualize the distribution of alcohol content

df['alcohol'].plot(kind='hist', bins=20, figsize=(8,5))

plt.title('Distribution of Alcohol Content')
plt.xlabel('Alcohol')
plt.ylabel('Number of Wines')
plt.show()
```

```python
# Compare average alcohol content across wine quality scores

alcohol_by_quality = df.groupby('quality')['alcohol'].mean()

alcohol_by_quality
```

```python
# Visualize the average alcohol content for each wine quality score

alcohol_by_quality.plot(kind='bar', figsize=(8,5))

plt.title('Average Alcohol Content by Wine Quality')
plt.xlabel('Quality Score')
plt.ylabel('Average Alcohol')
plt.xticks(rotation=0)
plt.show()
```

```python
# Compare average volatile acidity across wine quality scores

volatile_acidity_by_quality = df.groupby('quality')['volatile acidity'].mean()
volatile_acidity_by_quality
```

```python
# Visualize the average volatile acidity for each wine quality score

volatile_acidity_by_quality.plot(kind='bar', figsize=(8,5))

plt.title('Average Volatile Acidity by Wine Quality')
plt.xlabel('Quality Score')
plt.ylabel('Average Volatile Acidity')
plt.xticks(rotation=0)
plt.show()
```

```python
# Calculate correlations between numerical features and wine quality

correlation_with_quality = df.corr(numeric_only=True)['quality'].sort_values(ascending=False)
correlation_with_quality
```

```python
# Visualize correlations between numerical features and wine quality

correlation_with_quality.plot(kind='bar', figsize=(10,6))

plt.title('Correlation of Numerical Features with Wine Quality')
plt.xlabel('Features')
plt.ylabel('Correlation with Quality')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Visualize the distributions of the main numerical featurea

df[['fixed acidity', 'volatile acidity', 'citric acid', 'residual sugar', 'chlorides', 'sulphates', 'alcohol']].hist(figsize=(12,8), bins=20)

plt.suptitle('Distribution of key Wine Features')
plt.tight_layout()
plt.show()
```

### EDA Sammary

- **Quality Distribution:** Wine quality scores were concentrated mainly around 5 and 6, while relatively few wines received scores of 3 or 8.
- **Chemical Features:** The analysis also examined alcohol content and volatile acidity across different quality levels, helping identify patterns between chemical properties and wine quality.
- **Correlation Analysis:** Correlation analysis was used to identify numerical features that may have relationships with the quality score.
- **Feature Distributions:** Key chemical features were visulized to understand their ranges, patterns, and potential outliers.
- **Overall Findings:** These findings provide a foundation for feature selection and building the machine learning models.
